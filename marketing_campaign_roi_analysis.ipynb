```python
# Load the cleaned dataset in both Google Colab and the GitHub repository.
from pathlib import Path
import pandas as pd

project_root = Path.cwd()
repo_csv = project_root.parent / "data" / "marketing_campaign_roi_cleaned.csv"
colab_csv = Path("/content/marketing_campaign_roi_cleaned.csv")

if repo_csv.exists():
    csv_path = repo_csv
elif colab_csv.exists():
    csv_path = colab_csv
else:
    raise FileNotFoundError(
        "Could not find marketing_campaign_roi_cleaned.csv. "
        "In Google Colab, upload the CSV to /content. "
        "In the GitHub repository, keep it in data/."
    )

df = pd.read_csv(csv_path)
print(f"Loaded dataset from: {csv_path}")
```

```python
df["ROI"] = ((df["Revenue"] - df["Ad_Spend"])/ df["Ad_Spend"]) * 100
```

```python
df["ROAS"] = df["Revenue"] / df["Ad_Spend"]
```
